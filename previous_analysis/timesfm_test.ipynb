```python
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import timedelta
from sklearn.metrics import mean_absolute_error
from tqdm import tqdm

from timesfm import TimesFM_2p5_200M_torch, ForecastConfig  # ✅ 官方 2.5 API

print("✅ 已匯入 TimesFM_2p5_200M_torch 與 ForecastConfig")

# ==========================================
# 1. 全域參數
# ==========================================
HORIZON_LEN = 126           # 預測步數
DECAY_METHODS = ['ffill', 'geometric', 'log', 'cos']
TEST_START_DATE = "2024-01-01"   # 若你在 Cell 1 用別的，這裡改成一致
TOP_N = 5                         # 每次挑幾個外生變數

# 需要遞減的衝擊變數
SHOCK_VARS = ['FOMC_Sentiment', 'Mich_Sentiment', 'Financial_Cond', 'Inflation_Exp']

# ==========================================
# 2. 智能分流生成器 (Smart Decay)
# ==========================================
class SmartDecayGenerator:
    @staticmethod
    def generate_future_path(last_val, steps, col_name, method='cos'):
        """生成未來預測路徑: Level hold, Shock decay"""
        t = np.arange(steps)
        if col_name not in SHOCK_VARS:
            # 連續變數：直接 hold
            return np.full(steps, last_val)
        
        # 衝擊變數：遞減歸零
        if method == 'cos':
            return last_val * 0.5 * (1 + np.cos(t * np.pi / steps))
        elif method == 'geometric':
            return last_val * (0.95 ** t)
        elif method == 'log':
            return last_val / (1 + 0.1 * np.log1p(t))
        else:  # ffill
            return np.full(steps, last_val)

    @staticmethod
    def generate_sawtooth_history(series, date_map, col_name, method='cos'):
        """生成歷史鋸齒特徵: 遇到事件或值變化就 reset，之後 decay"""
        arr = series.values
        dates = series.index
        n = len(arr)
        
        # 連續變數不做鋸齒化
        if col_name not in SHOCK_VARS:
            return series
        
        decayed_arr = np.zeros(n)
        last_reset_val = arr[0]
        steps_since_reset = 0
        decay_span = 45   # 最大衰減窗
        event_dates = date_map.get(col_name, set())
        
        for i in range(n):
            current_date = dates[i]
            is_reset = (
                i == 0
                or (current_date in event_dates)
                or (arr[i] != arr[i-1])
            )
            if is_reset:
                last_reset_val = arr[i]
                steps_since_reset = 0
                decayed_arr[i] = last_reset_val
            else:
                steps_since_reset += 1
                if method != 'ffill':
                    if method == 'cos':
                        t = min(steps_since_reset, decay_span)
                        factor = 0.5 * (1 + np.cos(t * np.pi / decay_span))
                        decayed_arr[i] = last_reset_val * factor
                    elif method == 'geometric':
                        decayed_arr[i] = last_reset_val * (0.95 ** steps_since_reset)
                    elif method == 'log':
                        decayed_arr[i] = last_reset_val / (1 + 0.1 * np.log1p(steps_since_reset))
                else:
                    decayed_arr[i] = last_reset_val
        return pd.Series(decayed_arr, index=dates)

# ==========================================
# 3. 純 TimesFM 2.5 回測 (原生 covariates)
# ==========================================
def run_github_timesfm_experiment(df, target_col, date_map):
    print(f"\n>>> [Step 2] 啟動 GitHub TimesFM 2.5 (原生 dynamic_numerical_covariates)...")
    
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"    使用裝置: {device}")

    # 建立測試日期 (每月跑一次)
    test_dates = [d for d in df.index if d >= pd.to_datetime(TEST_START_DATE)]
    test_dates = test_dates[::30]
    if not test_dates:
        raise ValueError("測試日期集合為空，請檢查 TEST_START_DATE / 資料索引。")

    # 先用第一個測試日期選 feature，決定 covariate 維度
    sample_feats = select_top_features(df, target_col, test_dates[0], top_n=TOP_N)
    if not sample_feats:
        raise ValueError("select_top_features 沒有選出任何特徵，請檢查資料與設定。")
    num_cov = len(sample_feats)
    print(f"    Covariates 維度(示例): {num_cov}, features: {sample_feats}")

    # 1. 建 ForecastConfig
    config = ForecastConfig(
        horizon=HORIZON_LEN,
        num_dynamic_numerical_covariates=num_cov,
        frequency="D",
    )

    # 2. 初始化 TimesFM 2.5 模型並載入權重
    print("    正在載入 TimesFM 2.5 權重 (google/timesfm-2.5-200m-pytorch)...")
    tfm = TimesFM_2p5_200M_torch.from_pretrained("google/timesfm-2.5-200m-pytorch")
    tfm.to(device)
    tfm.compile(config)
    print("    模型載入與 compile 完成。")

    results = {m: [] for m in DECAY_METHODS}
    results['Date'] = []

    for t in tqdm(test_dates, desc="Rolling"):
        try:
            future_true = df.loc[t:].head(HORIZON_LEN)[target_col].values
            if len(future_true) < HORIZON_LEN:
                continue
            
            feats = select_top_features(df, target_col, t, top_n=TOP_N)
            if not feats:
                continue
            
            history = df.loc[:t - timedelta(days=1)]
            context_y = history[target_col].values.astype(np.float32)

            for method in DECAY_METHODS:
                cov_list = []
                for f in feats:
                    # 歷史段：鋸齒化
                    hist_part = SmartDecayGenerator.generate_sawtooth_history(
                        history[f], date_map, f, method
                    ).values
                    # 未來段：Smart Decay / Hold
                    last_v = history[f].iloc[-1]
                    fut_part = SmartDecayGenerator.generate_future_path(
                        last_v, HORIZON_LEN, f, method
                    )
                    full_seq = np.concatenate([hist_part, fut_part])
                    cov_list.append(full_seq)
                
                full_cov_arr = np.stack(cov_list, axis=1)  # (T_hist+H, F)
                
                # TimesFM context + horizon 對齊
                input_len = 512
                if len(context_y) > input_len:
                    context_in = context_y[-input_len:]
                    cov_in = full_cov_arr[-(input_len + HORIZON_LEN):]
                else:
                    context_in = context_y
                    cov_in = full_cov_arr[-(len(context_in) + HORIZON_LEN):]

                # y: (1, T_ctx)
                y = context_in[np.newaxis, :].astype(np.float32)
                # cov: (1, T_ctx+H, F)
                cov = cov_in[np.newaxis, :, :].astype(np.float32)

                y_t = torch.from_numpy(y).to(device)
                cov_t = torch.from_numpy(cov).to(device)

                # 3. 原生 covariates 預測
                out = tfm.forecast(
                    y=y_t,
                    dynamic_numerical_covariates=cov_t,
                )
                # point_forecast 通常 shape = (B, H)
                pred = out.point_forecast[0].detach().cpu().numpy()
                
                mae = mean_absolute_error(future_true, pred)
                results[method].append(mae)

            results['Date'].append(t)
        
        except Exception as e:
            # Debug 用：需要時打開
            # print(f"Error at {t}: {e}")
            continue

    if not results['Date']:
        return pd.DataFrame()

    return pd.DataFrame(results).set_index('Date')

# ==========================================
# 4. 執行實驗
# ==========================================
if 'data_pool' in locals() and 'EVENT_DATES_MAP' in locals():
    df_res = run_github_timesfm_experiment(data_pool, TARGET_COL, EVENT_DATES_MAP)
    
    if not df_res.empty:
        print("\n📊 === 純 TimesFM 2.5 最終結果 (Smart Decay + 原生 covariates) ===")
        print(df_res.mean().sort_values().to_string())
        
        means = df_res.mean()
        if means.max() - means.min() < 1e-6:
            print("\n⚠️ 警告：各 decay 方法 MAE 幾乎一樣，TimesFM 可能對外生變數不敏感或權重極低。")
        else:
            df_res.plot(figsize=(12, 6), linewidth=2)
            plt.title("Rolling MAE: Pure TimesFM 2.5 with Smart Decay Covariates")
            plt.ylabel("MAE")
            plt.show()
    else:
        print("❌ 無結果（可能所有滾動視窗都在 try/except 裡被跳過）")
else:
    print("❌ 請先在 Cell 1 準備 data_pool / EVENT_DATES_MAP / TARGET_COL / select_top_features")
```

```python
import torch
import timesfm
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import timedelta
from sklearn.metrics import mean_absolute_error
from tqdm import tqdm

# ==========================================
# 5. 智能分流生成器 (Smart Decay)
# ==========================================
class SmartDecayGenerator:
    @staticmethod
    def generate_future_path(last_val, steps, col_name, method='cos'):
        """
        ★ 防偷看核心 ★
        1. 輸入只有 last_val (歷史最後一天的值)。
        2. 輸出完全依賴數學公式。
        3. 根據 col_name 決定是 Level (FFill) 還是 Shock (Decay)。
        """
        t = np.arange(steps)
        
        # --- 規則 A: 連續性變數 (Level) ---
        # 油價、股價、殖利率 -> 強制補前值
        if col_name in CONTINUOUS_VARS:
            return np.full(steps, last_val)
            
        # --- 規則 B: 非連續性變數 (Shock) ---
        # CPI、FOMC -> 允許遞減歸零
        else:
            if method == 'cos':
                return last_val * 0.5 * (1 + np.cos(t * np.pi / steps))
            elif method == 'geometric':
                return last_val * (0.95 ** t)
            elif method == 'log':
                return last_val / (1 + 0.1 * np.log1p(t))
            else: # ffill (對照組)
                return np.full(steps, last_val)

    @staticmethod
    def generate_sawtooth_history(series, date_map, col_name, method='cos'):
        """
        歷史特徵工程: 重現過去的已知狀態
        """
        arr = series.values
        dates = series.index
        n = len(arr)
        
        # 連續變數不需要做鋸齒處理 (因為每天都在變)
        if col_name in CONTINUOUS_VARS:
            return series
            
        decayed_arr = np.zeros(n)
        last_reset_val = arr[0]
        steps_since_reset = 0
        decay_span = 45 
        event_dates = date_map.get(col_name, set())
        
        for i in range(n):
            current_date = dates[i]
            is_reset = (i == 0) or (current_date in event_dates) or (arr[i] != arr[i-1])
            
            if is_reset:
                last_reset_val = arr[i]
                steps_since_reset = 0
                decayed_arr[i] = last_reset_val
            else:
                steps_since_reset += 1
                if method != 'ffill':
                    if method == 'cos':
                        t = min(steps_since_reset, decay_span)
                        factor = 0.5 * (1 + np.cos(t * np.pi / decay_span))
                        decayed_arr[i] = last_reset_val * factor
                    elif method == 'geometric':
                        decayed_arr[i] = last_reset_val * (0.95 ** steps_since_reset)
                    elif method == 'log':
                        decayed_arr[i] = last_reset_val / (1 + 0.1 * np.log1p(steps_since_reset))
                else:
                    decayed_arr[i] = last_reset_val
                    
        return pd.Series(decayed_arr, index=dates)

# ==========================================
# 6. 回測執行
# ==========================================
HORIZON_LEN = 126
DECAY_METHODS = ['ffill', 'geometric', 'log', 'cos']

def run_pure_timesfm_experiment(df, target_col, date_map):
    print(f"\n>>> [Step 2] 開始純 TimesFM 回測 (智能分流版)...")
    
    device = "cuda" if torch.cuda.is_available() else "cpu"
    # 載入模型
    tfm = timesfm.TimesFM_2p5_200M_torch.from_pretrained(
        "google/timesfm-2.5-200m-pytorch", 
        device=device
    )
    
    # 編譯設定
    config = timesfm.ForecastConfig(
        max_context=2048, 
        max_horizon=HORIZON_LEN+10, 
        normalize_inputs=True
    )
    tfm.compile(config)
    
    test_dates = [d for d in df.index if d >= pd.to_datetime(TEST_START_DATE)]
    test_dates = test_dates[::30] 
    
    results = {m: [] for m in DECAY_METHODS}
    results['Date'] = []
    
    for t in tqdm(test_dates, desc="Rolling"):
        try:
            future_true = df.loc[t:].head(HORIZON_LEN)[target_col].values
            if len(future_true) < HORIZON_LEN: continue
            
            # 1. 選股 (Top 5 + FOMC)
            # ★ 關鍵：只用 t 以前的資料選股 ★
            feats = select_top_features(df, target_col, t, top_n=TOP_N)
            
            # 2. 準備主序列
            history = df.loc[:t - timedelta(days=1)]
            context_y = history[target_col].values.astype(np.float32)
            
            # 3. 測試遞減法
            for method in DECAY_METHODS:
                cov_list = []
                for f in feats:
                    # A. 歷史部分 (History)
                    hist_part = SmartDecayGenerator.generate_sawtooth_history(
                        history[f], date_map, f, method
                    ).values
                    
                    # B. 未來部分 (Future)
                    # ★ 關鍵：只用 history 的最後一個值，憑空畫出未來 ★
                    last_v = history[f].iloc[-1]
                    fut_part = SmartDecayGenerator.generate_future_path(
                        last_v, HORIZON_LEN, f, method
                    )
                    
                    # 拼接
                    full_seq = np.concatenate([hist_part, fut_part])
                    cov_list.append(full_seq)
                
                # 轉置 & 截取 (符合 TimesFM 格式)
                full_cov_arr = np.stack(cov_list, axis=1)
                
                # 簡單截取最後 512 點 Context + 126 點 Horizon
                input_len = 512
                if len(context_y) > input_len:
                    context_in = context_y[-input_len:]
                    cov_in = full_cov_arr[-(input_len + HORIZON_LEN):]
                else:
                    context_in = context_y
                    cov_in = full_cov_arr
                
                cov_tensor = torch.tensor(cov_in).float().unsqueeze(0).to(device)
                
                # C. 預測
                # 嘗試使用正確的參數名稱
                try:
                    # TimesFM 2.5 官方推薦參數名
                    forecast_result = tfm.forecast(
                        inputs=[context_in],
                        dynamic_numerical_covariates=cov_tensor,
                        horizon=HORIZON_LEN
                    )
                except TypeError:
                    # 舊版參數名
                    try:
                        forecast_result = tfm.forecast(
                            inputs=[context_in],
                            dynamic_covariates=cov_tensor,
                            horizon=HORIZON_LEN
                        )
                    except TypeError:
                        # 真的不支援
                        if len(results['Date']) == 0 and method == DECAY_METHODS[0]:
                            print("❌ 警告: TimesFM 不接受外生變數")
                        forecast_result = tfm.forecast(inputs=[context_in], horizon=HORIZON_LEN)

                pred = forecast_result[0][0]
                if isinstance(pred, torch.Tensor): pred = pred.cpu().numpy()
                
                mae = mean_absolute_error(future_true, pred)
                results[method].append(mae)
            
            results['Date'].append(t)
            
        except Exception: continue
            
    return pd.DataFrame(results).set_index('Date')

# 執行
if 'data_pool' in locals() and 'EVENT_DATES_MAP' in locals():
    df_res = run_pure_timesfm_experiment(data_pool, TARGET_COL, EVENT_DATES_MAP)
    
    if not df_res.empty:
        print("\n📊 === 最終結果: 智能分流 (2024-Now) ===")
        print(df_res.mean().sort_values().to_string())
        
        # 檢查是否有效
        means = df_res.mean()
        if means.max() - means.min() < 1e-6:
            print("\n⚠️ 警告：所有方法結果相同！")
            print("   原因：TimesFM 對外生變數反應冷淡，或降級為單變量。")
        else:
            df_res.plot(figsize=(12, 6), linewidth=2)
            plt.title("Rolling MAE: Continuous(Hold) vs Discontinuous(Decay)")
            plt.ylabel("MAE")
            plt.show()
    else:
        print("❌ 無結果")
else:
    print("❌ 請先執行 Cell 1")
```

```python
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import timedelta
from sklearn.metrics import mean_absolute_error
from tqdm import tqdm

from timesfm import TimesFM_2p5_200M_torch, ForecastConfig  # ✅ 官方 2.5 API

print("✅ 已匯入 TimesFM_2p5_200M_torch 與 ForecastConfig")

# ==========================================
# 1. 全域參數
# ==========================================
HORIZON_LEN = 126
DECAY_METHODS = ['ffill', 'geometric', 'log', 'cos']
TEST_START_DATE = "2024-01-01"
TOP_N = 5

SHOCK_VARS = ['FOMC_Sentiment', 'Mich_Sentiment', 'Financial_Cond', 'Inflation_Exp']

# ==========================================
# 2. 智能分流生成器
# ==========================================
class SmartDecayGenerator:
    @staticmethod
    def generate_future_path(last_val, steps, col_name, method='cos'):
        t = np.arange(steps)
        if col_name not in SHOCK_VARS:
            return np.full(steps, last_val)
        if method == 'cos':
            return last_val * 0.5 * (1 + np.cos(t * np.pi / steps))
        elif method == 'geometric':
            return last_val * (0.95 ** t)
        elif method == 'log':
            return last_val / (1 + 0.1 * np.log1p(t))
        else:
            return np.full(steps, last_val)

    @staticmethod
    def generate_sawtooth_history(series, date_map, col_name, method='cos'):
        arr = series.values
        dates = series.index
        n = len(arr)
        if col_name not in SHOCK_VARS:
            return series
        
        decayed_arr = np.zeros(n)
        last_reset_val = arr[0]
        steps_since_reset = 0
        decay_span = 45
        event_dates = date_map.get(col_name, set())
        
        for i in range(n):
            current_date = dates[i]
            is_reset = (
                i == 0
                or (current_date in event_dates)
                or (arr[i] != arr[i-1])
            )
            if is_reset:
                last_reset_val = arr[i]
                steps_since_reset = 0
                decayed_arr[i] = last_reset_val
            else:
                steps_since_reset += 1
                if method != 'ffill':
                    if method == 'cos':
                        t = min(steps_since_reset, decay_span)
                        factor = 0.5 * (1 + np.cos(t * np.pi / decay_span))
                        decayed_arr[i] = last_reset_val * factor
                    elif method == 'geometric':
                        decayed_arr[i] = last_reset_val * (0.95 ** steps_since_reset)
                    elif method == 'log':
                        decayed_arr[i] = last_reset_val / (1 + 0.1 * np.log1p(steps_since_reset))
                else:
                    decayed_arr[i] = last_reset_val
        return pd.Series(decayed_arr, index=dates)

# ==========================================
# 3. 純 TimesFM 2.5 回測 (原生 covariates)
# ==========================================
def run_github_timesfm_experiment(df, target_col, date_map):
    print(f"\n>>> [Step 2] 啟動 GitHub TimesFM 2.5 (原生 dynamic_numerical_covariates)...")
    
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"    使用裝置: {device}")

    test_dates = [d for d in df.index if d >= pd.to_datetime(TEST_START_DATE)]
    test_dates = test_dates[::30]
    if not test_dates:
        raise ValueError("測試日期集合為空，請檢查 TEST_START_DATE / 資料索引。")

    sample_feats = select_top_features(df, target_col, test_dates[0], top_n=TOP_N)
    if not sample_feats:
        raise ValueError("select_top_features 沒有選出任何特徵，請檢查資料與設定。")
    num_cov = len(sample_feats)
    print(f"    Covariates 維度(示例): {num_cov}, features: {sample_feats}")

    config = ForecastConfig(
        max_context=512,
        max_horizon=HORIZON_LEN,
        normalize_inputs=True,
    )

    print("    正在載入 TimesFM 2.5 權重 (google/timesfm-2.5-200m-pytorch)...")
    tfm = TimesFM_2p5_200M_torch.from_pretrained("google/timesfm-2.5-200m-pytorch")
    tfm.compile(config)
    print("    模型載入與 compile 完成。")

    results = {m: [] for m in DECAY_METHODS}
    results['Date'] = []

    for t in tqdm(test_dates, desc="Rolling"):
        try:
            future_true = df.loc[t:].head(HORIZON_LEN)[target_col].values
            if len(future_true) < HORIZON_LEN:
                continue
            
            feats = select_top_features(df, target_col, t, top_n=TOP_N)
            if not feats:
                continue
            
            history = df.loc[:t - timedelta(days=1)]
            context_y = history[target_col].values.astype(np.float32)

            for method in DECAY_METHODS:
                cov_list = []
                for f in feats:
                    hist_part = SmartDecayGenerator.generate_sawtooth_history(
                        history[f], date_map, f, method
                    ).values
                    last_v = history[f].iloc[-1]
                    fut_part = SmartDecayGenerator.generate_future_path(
                        last_v, HORIZON_LEN, f, method
                    )
                    full_seq = np.concatenate([hist_part, fut_part])
                    cov_list.append(full_seq)
                
                full_cov_arr = np.stack(cov_list, axis=1)  # (T_hist+H, F)
                
                input_len = 512
                if len(context_y) > input_len:
                    context_in = context_y[-input_len:]
                    cov_in = full_cov_arr[-(input_len + HORIZON_LEN):]
                else:
                    context_in = context_y
                    cov_in = full_cov_arr[-(len(context_in) + HORIZON_LEN):]

                y = context_in[np.newaxis, :].astype(np.float32)    # (1, T_ctx)
                cov = cov_in[np.newaxis, :, :].astype(np.float32)   # (1, T_ctx+H, F)

                out = tfm.forecast(
                    y=y,
                    dynamic_numerical_covariates=cov,
                )
                pred = out.point_forecast[0]  # (H,)

                mae = mean_absolute_error(future_true, pred)
                results[method].append(mae)

            results['Date'].append(t)

        except Exception as e:
            print(f"Error at {t}: {type(e).__name__} -> {e}")
            break

    if not results['Date']:
        return pd.DataFrame()

    return pd.DataFrame(results).set_index('Date')

# ==========================================
# 4. 執行實驗
# ==========================================
if 'data_pool' in locals() and 'EVENT_DATES_MAP' in locals():
    df_res = run_github_timesfm_experiment(data_pool, TARGET_COL, EVENT_DATES_MAP)
    
    if not df_res.empty:
        print("\n📊 === 純 TimesFM 2.5 最終結果 (Smart Decay + 原生 covariates) ===")
        print(df_res.mean().sort_values().to_string())
        
        means = df_res.mean()
        if means.max() - means.min() < 1e-6:
            print("\n⚠️ 警告：各 decay 方法 MAE 幾乎一樣，TimesFM 可能對外生變數不敏感或權重極低。")
        else:
            df_res.plot(figsize=(12, 6), linewidth=2)
            plt.title("Rolling MAE: Pure TimesFM 2.5 with Smart Decay Covariates")
            plt.ylabel("MAE")
            plt.show()
    else:
        print("❌ 無結果（可能所有滾動視窗都在 try/except 裡被跳過）")
else:
    print("❌ 請先在 Cell 1 準備 data_pool / EVENT_DATES_MAP / TARGET_COL / select_top_features")
```

```python
import numpy as np
import pandas as pd
from datetime import timedelta
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from tqdm import tqdm
import matplotlib.pyplot as plt

from timesfm import TimesFM_2p5_200M_torch, ForecastConfig  # 你這個環境的 TimesFM

# ==========================================
# 1. 全域設定
# ==========================================
TARGET_COL = 'Target_DGS2'
TEST_START_DATE = '2022-01-01'   # 驗證期起點
LOOKBACK_WINDOW = 12             # 特徵選擇用最近 12 個月
CORR_THRESHOLD = 0.3             # Diff 相關門檻
HORIZON_LEN = 126                # 每次預測 horizon

# ==========================================
# 2. 特徵選擇（沿用你 diff 相關邏輯）
# ==========================================
def select_features_for_hybrid(df, target_col, test_start_date, threshold):
    valid_data = df[df.index < pd.to_datetime(test_start_date)].copy()
    monthly_data = valid_data.resample('ME').last().iloc[-LOOKBACK_WINDOW:]
    monthly_diff = monthly_data.diff().dropna()
    
    selected = []
    for col in df.columns:
        if col == target_col:
            continue
        if col in monthly_diff.columns:
            corr = monthly_diff[target_col].corr(monthly_diff[col])
            if not np.isnan(corr) and abs(corr) >= threshold:
                selected.append(col)
    if 'FOMC_Sentiment' in df.columns and 'FOMC_Sentiment' not in selected:
        selected.append('FOMC_Sentiment')
    return selected

# ==========================================
# 3. Hybrid: XReg + TimesFM Residual
# ==========================================
def run_timesfm_hybrid_xreg(df, target_col):
    print("\n>>> 啟動 TimesFM Hybrid (XReg + TimesFM Residual)...")

    # 決定全局外生變數清單（用訓練期資料）
    feats = select_features_for_hybrid(df, target_col, TEST_START_DATE, CORR_THRESHOLD)
    if not feats:
        raise ValueError("沒有任何外生變數達到相關門檻，請降低 CORR_THRESHOLD 或檢查資料。")
    print(f"   使用的外生變數: {feats}")

    # TimesFM 設定
    config = ForecastConfig(
        max_context=512,
        max_horizon=HORIZON_LEN,
        normalize_inputs=True,
    )
    print("   載入 TimesFM 2.5 權重...")
    tfm = TimesFM_2p5_200M_torch.from_pretrained("google/timesfm-2.5-200m-pytorch")
    tfm.compile(config)
    print("   TimesFM 準備完成。")

    # XReg (Ridge)
    ridge = Ridge(alpha=10.0, fit_intercept=True)
    scaler = StandardScaler()

    # 建立 rolling 測試日期：驗證從 TEST_START_DATE 開始，每約 30 天一個點
    test_dates_all = [d for d in df.index if d >= pd.to_datetime(TEST_START_DATE)]
    test_dates = test_dates_all[::30]
    if not test_dates:
        raise ValueError("測試日期集合為空，請檢查 TEST_START_DATE / 資料索引。")

    print(f"   訓練資料起點: {df.index.min().date()}，"
          f"每次滾動訓練期為 [2015-01-01, t-1]，"
          f"驗證起點: {pd.to_datetime(TEST_START_DATE).date()}，"
          f"共 {len(test_dates)} 個滾動點，每次預測 {HORIZON_LEN} 天。")

    pure_tfm_mae_list = []
    hybrid_mae_list = []
    date_list = []

    input_len = 512  # TimesFM context 上限

    for t in tqdm(test_dates, desc="Rolling Hybrid"):
        # 未來真實值 (驗證期每個 t 的 y_true)
        future_true = df.loc[t:].head(HORIZON_LEN)[target_col].values
        if len(future_true) < HORIZON_LEN:
            continue

        # 訓練期：從 2015-01-01 到 t-1（data_pool 一開始就是從 2015-01-01 建的）
        train_df = df[df.index < t]
        if len(train_df) < 200:
            continue

        # ===== XReg: 用外生變數直接預測 y =====
        X_train_raw = train_df[feats].values.astype(np.float32)
        y_train = train_df[target_col].values.astype(np.float32)

        X_train = scaler.fit_transform(X_train_raw)
        ridge.fit(X_train, y_train)

        # 未來 X：簡化成 hold 最後一天的外生變數
        last_row = df.loc[:t].iloc[-1]
        X_future_raw = np.tile(last_row[feats].values.astype(np.float32), (HORIZON_LEN, 1))
        X_future = scaler.transform(X_future_raw)
        ridge_pred_future = ridge.predict(X_future)  # (H,)

        # 訓練期的 XReg 預測，用來算 residual
        ridge_pred_train = ridge.predict(X_train)    # (len(train_df),)
        resid_train = y_train - ridge_pred_train     # (len(train_df),)

        # ===== TimesFM: 預測 residual 未來值 =====
        context_resid = resid_train.astype(np.float32)
        if len(context_resid) > input_len:
            context_in = context_resid[-input_len:]
        else:
            context_in = context_resid
        inputs_list = [context_in]

        point_forecast_resid, _ = tfm.forecast(
            inputs=inputs_list,
            freq=[0],                # 0 = daily
            window_size=HORIZON_LEN
        )
        tfm_resid_pred_future = point_forecast_resid[0]  # (H,)

        # 最終 Hybrid 預測 = XReg(y) + TimesFM(residual)
        final_pred = ridge_pred_future + tfm_resid_pred_future

        # 同時算一次純 TimesFM 預測 y 的 MAE 做對照
        context_y = train_df[target_col].values.astype(np.float32)
        if len(context_y) > input_len:
            context_y_in = context_y[-input_len:]
        else:
            context_y_in = context_y
        pure_inputs = [context_y_in]
        pure_point_forecast, _ = tfm.forecast(
            inputs=pure_inputs,
            freq=[0],
            window_size=HORIZON_LEN
        )
        pure_tfm_pred = pure_point_forecast[0]

        # 評估
        pure_mae = mean_absolute_error(future_true, pure_tfm_pred)
        hybrid_mae = mean_absolute_error(future_true, final_pred)

        pure_tfm_mae_list.append(pure_mae)
        hybrid_mae_list.append(hybrid_mae)
        date_list.append(t)

    if not date_list:
        print("❌ 無有效 rolling 結果，請檢查資料長度或 TEST_START_DATE。")
        return pd.DataFrame()

    res_df = pd.DataFrame(
        {
            "Pure_TimesFM_MAE": pure_tfm_mae_list,
            "Hybrid_XReg_TimesFM_MAE": hybrid_mae_list,
        },
        index=pd.to_datetime(date_list),
    )
    return res_df

# ==========================================
# 4. 執行 Hybrid 並檢視結果
# ==========================================
res_df = run_timesfm_hybrid_xreg(data_pool, TARGET_COL)

if not res_df.empty:
    print("\n📊 Rolling MAE 平均值：")
    print(res_df.mean().to_string())

    ax = res_df.plot(figsize=(12, 6), marker='o')
    ax.set_title("Rolling MAE: Pure TimesFM vs Hybrid (XReg + TimesFM Residual)")
    ax.set_ylabel("MAE")
    plt.show()
else:
    print("❌ 無結果（所有滾動點都被篩掉）")
```
